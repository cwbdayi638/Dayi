```python
import pygmt
import pandas as pd

# Check PyGMT version
print(pygmt.__version__)

# Example plot
fig = pygmt.Figure()
fig.coast(projection="N15c", region="g", frame=True, land="tan", water="lightblue")
fig.text(position="MC", text="PyGMT", font="80p,Helvetica-Bold,red@75")
fig.show()
```

```python
# prompt: plot taiwan

fig = pygmt.Figure()
fig.coast(region="TW", projection="M15c", land="brown", water="lightblue", shorelines=True)
fig.basemap(frame=["a", "wsen"])
fig.show()
```

```python
# prompt: plot taiwan with topography

# Define region for Taiwan
region = [119, 123, 21.5, 26]

# Create a Figure instance
fig = pygmt.Figure()

# Add a base map with topography data
fig.grdimage(
    grid="@earth_relief_03m",
    region=region,
    projection="M15c",
    shading=True,  # Add shading for topography
    cmap="geo",  # Use a geographic colormap
)

# Add coastlines
fig.coast(
    region=region,
    projection="M15c",
    shorelines=True,
    frame=["a", "wsen"]
)

# Show the plot
fig.show()
```
